```python
def bill_analyzer(units):
    bill = 0
    #First 100 units
    if units <= 100:
        bill = units * 3
    #101–200 units
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    #Above 200 units
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)
    #Usage Classification
    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"
    return bill, status
units_consumed = int(input("Enter units consumed: "))
total_bill, usage_status = bill_analyzer(units_consumed)
print("Enter units consumed:",units_consumed)
print("Total Bill: ₹", total_bill)
print("Usage Level:", usage_status)
```
