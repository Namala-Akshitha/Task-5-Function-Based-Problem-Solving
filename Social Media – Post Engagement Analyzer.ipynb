```python
def socialmedia_post(likes_list):
    total_likes = 0
    #Loop to calculate total likes
    for like in likes_list:
        total_likes += like
    print("Total Likes:", total_likes)
    #Condition check
    if total_likes >= 1000:
        return "Viral Post"
    else:
        return "Normal Engagement"
#Example usage
likes = [120, 250, 300, 150, 200]   #likes from different users/sources
result=socialmedia_post(likes)
print("Post Status:", result)
```
